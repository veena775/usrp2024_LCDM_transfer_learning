```python
Pk_type = 'MPk'
cosm_type = 'nwLH'
n_sims_nwLH = [200, 400, 800, 1600, 2000]
n_sims_BSQ = [0, 2000, 4000, 8000, 16000, 32000]

minimum = np.array([0.1, 0.02, 0.50, 0.80, 0.60, 0.01])
maximum = np.array([0.5, 0.08, 0.90, 1.20, 1.00, 1.0])
colors = ['C0', 'C1', 'C2', 'C3', 'C4', 'C5']
num_params = 6

for color_index, n_BSQ in enumerate(n_sims_BSQ):
    mse_per_BSQ = []
    n_per_BSQ = []
    avg_mse = []
    for n_nwLH in n_sims_nwLH:
        n = []
        mse = []
        for n_model in range(10):
            name = 'transfer10_network1_'+str(n_nwLH)+'_nwLH_'+str(n_BSQ)+'_BSQ'
            test = np.loadtxt('results/Results_test_'+str(Pk_type)+'_'+str(cosm_type)+'_params_'+str(name)+'model'+str(n_model)+'.txt')
            
            test_true = test[:, :num_params]     # First half (true values)
            test_predicted = test[:, num_params:]  # Second half (predicted values)
    
            test_predicted = test_predicted * (maximum - minimum) + minimum
            test_true = test_true * (maximum - minimum) + minimum
    
            mse.append(mean_squared_error(test_true, test_predicted))
            n.append(n_nwLH)
    
        avg_mse.append(np.mean(mse))
        mse_per_BSQ.append(mse)
        n_per_BSQ.append(n)
        
    mse_per_BSQ = np.array(mse_per_BSQ).flatten()
    n_per_BSQ = np.array(n_per_BSQ).flatten()
    plt.scatter(n_per_BSQ, mse_per_BSQ, label = 'n BSQ sim = '+str(n_BSQ), s =5, color = colors[color_index])
    plt.plot(n_sims_nwLH, avg_mse, color = colors[color_index], marker = 'o')

plt.legend()
plt.show()

        
```
