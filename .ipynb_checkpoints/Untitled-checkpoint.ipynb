## Learning rate scatter plot

```python
from matplotlib.lines import Line2D

# Load study
study_name = "Pk_nwLH_params_transfer10_network2_50_nwLH_22000_BSQ_fixed_ft_test"
storage = "sqlite:///nwLH.db"
study = optuna.load_study(study_name=study_name, storage=storage)

# Gather data
trial_nums = []
lrs = []
colors = []

completed_trials_id  = []
pruned_trials_id = []

best_trial = study.best_trial
best_lr = best_trial.params.get("lr", None)
best_trial_number = best_trial.number

for trial in study.trials:
    lr = trial.params.get("lr", None)
    if lr is not None:
        trial_nums.append(trial.number)
        lrs.append(lr)
        if trial.state.name == "COMPLETE":
            colors.append("red" if trial.number == best_trial_number else "blue")
            completed_trials_id.append(trial.number)
        elif trial.state.name == "PRUNED":
            colors.append("lightgray")
            pruned_trials_id.append(trial.number)

# Plot
plt.figure(figsize=(10, 6))
for t, lr, c in zip(trial_nums, lrs, colors):
    plt.scatter(t, lr, color=c, s=60)

# plt.yscale("log")
plt.xlabel("Trial Number", fontsize=14)
plt.ylabel("Learning Rate", fontsize=14)
plt.title(r"Optuna Trial Learning Rates: $M_{\nu} (P)$, 50 sims")

legend_elements = [
    Line2D([0], [0], marker='o', color='w', label=f'Best Trial (lr={best_lr:.1e})', markerfacecolor='red', markersize=10),
    Line2D([0], [0], marker='o', color='w', label='Completed trial', markerfacecolor='blue', markersize=10),
    Line2D([0], [0], marker='o', color='w', label='Pruned trial', markerfacecolor='lightgray', markersize=10)
]
plt.legend(handles=legend_elements)
plt.tight_layout()
plt.show()


print ('Completed trials: ', len(completed_trials_id))
print('Pruned trials: ', len(pruned_trials_id))
```

```python
import numpy as np
import os
import matplotlib.pyplot as plt
import optuna
from matplotlib.lines import Line2D

# === CONFIG ===
n_sims = 1400
Pk_type = 'MPk'

mother = '/scratch/network/vk9342/USRP2024_scratch/pytorch/'+Pk_type+'_nwLH/transfer10_network2_'+str(n_sims)+'_nwLH_22000_BSQ_fixed_ft_test/'
loss_dir = os.path.join(mother, 'losses')
storage = 'sqlite:///nwLH.db'
study_name = Pk_type+'_nwLH_params_transfer10_network2_'+str(n_sims)+'_nwLH_22000_BSQ_fixed_ft_test'

# === LOAD STUDY ===
study = optuna.load_study(study_name=study_name, storage=storage)
trial_status_dict = {trial.number: trial.state for trial in study.trials}

# === PLOT ===
plt.figure(figsize=(12, 6))
trial_nums = []
completed_trials_id  = []
pruned_trials_id = []
for fname in os.listdir(loss_dir):
    if fname.startswith('loss_') and fname.endswith('.txt'):
        trial_id = int(fname.split('_')[1].split('.')[0])
        data = np.loadtxt(os.path.join(loss_dir, fname))
        if data.ndim == 1:
            data = data.reshape(1, -1)

        epochs = data[:, 0]
        val_loss = data[:, 1]

        # Choose color based on trial status
        state = trial_status_dict.get(trial_id, None)
        if state == optuna.trial.TrialState.COMPLETE:
            color = 'blue'
            completed_trials_id.append(trial_id)
        elif state == optuna.trial.TrialState.PRUNED:
            color = 'red'
            pruned_trials_id.append(trial_id)
            
        # else:
        #     color = 'gray'  # fallback

        plt.plot(epochs, val_loss, alpha=0.4, color=color, label=f"Trial {trial_id}")
        trial_nums.append(trial_id)

# === LEGEND ===
legend_elements = [
    Line2D([0], [0], color='blue', lw=2, label='Completed trials'),
    Line2D([0], [0], color='red', lw=2, label='Pruned trials'),
    # Line2D([0], [0], color='gray', lw=2, label='Unknown status')
]
plt.legend(handles=legend_elements)

# === LABELS ===
plt.xlabel('Epoch')
plt.ylabel('Validation Loss')
plt.title(r'1400 sims, $M_{\nu}$ (MP)')
plt.grid(True)
plt.tight_layout()
plt.show()

print("Trials being plotted:", trial_nums)
print ('Completed trials: ', len(completed_trials_id))
print('Pruned trials: ', len(pruned_trials_id))
```
