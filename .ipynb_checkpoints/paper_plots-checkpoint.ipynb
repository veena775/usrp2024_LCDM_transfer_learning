```python
import numpy as np
import sys, os, time
import torch
import torch.nn as nn
import data, architecture
import optuna
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
import matplotlib as mpl

mpl.rcParams['mathtext.fontset'] = 'stix'
mpl.rcParams['font.family'] = 'serif'
```

## Get top 10 models

```python
################################### INPUT ############################################
# data parameters
f_Pk_norm = None #file with Pk to normalize Pk
seed      = 42       #seed to split data in train/valid/test
mode      = 'test'   #'train','valid','test' or 'all'

cosm_type = 'EQ'
additional_extension = '_fixed'   #FIXED TEST/VALID SET TO 300
params_ext = ''
Pk_type = 'Pk'
model_type = 'dynamic' #or 'dynamic' 'dynamic_fixed_final'
output_size = 6

n_sims_cosm = [25, 50, 75, 100, 125, 150, 175, 200, 400]
n_sims_BSQ = [2000, 4000, 8000, 16000, 22000]
```

```python
for n_BSQ in n_sims_BSQ: 
    for n_sims in n_sims_cosm:
        n = []
        mse = []

        name = 'transfer10_network2_'+str(n_sims)+'_'+cosm_type+'_'+str(n_BSQ)+'_BSQ' + additional_extension
        f_params  = '../real_params/'+str(cosm_type)+'_params'+params_ext+'.txt' 
        
        if n_BSQ == 0:
            f_Pk      = 'Pk_files/'+'all_'+str(Pk_type)+'_'+str(cosm_type)+params_ext+'.npy'
            input_size  = 79
            log = True
            
        
        else:
            extension = 'fhl10'
            f_Pk      = 'Pk_files/'+'all_'+cosm_type+'_'+str(n_BSQ)+'_BSQ_'+Pk_type+params_ext+'_'+str(extension)+'_fortransfer'+additional_extension+'.npy'
            input_size  = 10
            log = False
            

        study_name = str(Pk_type)+'_'+str(cosm_type)+'_params_'+str(name)
        mother = '/scratch/network/vk9342/USRP2024_scratch/pytorch/'+str(Pk_type)+'_'+str(cosm_type)+'/'+str(name)+'/models/'

        print(name)
        print(f_params)
        print(f_Pk)
        
    
        # architecture parameters
        # training parameters
        batch_size = 32
        # optuna parameters
        storage    = 'sqlite:///nwLH.db'
        
        
        
        ######################################################################################
        ######################################################################################
        # use GPUs if available
        if torch.cuda.is_available():
            print("CUDA Available")
            device = torch.device('cuda')
        else:
            print('CUDA Not Available')
            device = torch.device('cpu')
        
        # load the optuna study
        study = optuna.load_study(study_name=study_name, storage=storage)
        
        # get the scores of the study trials
        values = np.zeros(len(study.trials))
        completed = 0
        
        for i,t in enumerate(study.trials):
            values[i] = t.value
            if t.value is not None:  completed += 1
        
        # get the info of the best trial
        indexes = np.argsort(values)
        for indx in range(10):
            trial = study.trials[indexes[indx]]
            fout = 'results/Results_'+mode+'_'+str(Pk_type)+'_'+str(cosm_type)+'_params_'+str(name)+params_ext+'_model'+str(indx)+'.txt'
           
            # print("\nTrial number {}".format(trial.number))
            # print("Value: %.5e"%trial.value)
            for key, value in trial.params.items():
                lr       = trial.params['lr']
                wd       = trial.params['wd']
                n_layers       = trial.params['n_layers']
                p       = trial.params['dropout_l']
                
            fmodel = mother +'model_%d.pt'%trial.number

            
        
            # # generate the architecture
            if model_type == 'dynamic_fixed_final':
                out_fs = [trial.params[f'n_units_l{i}'] for i in range(n_layers-1)]
                model = architecture.dynamic_model_fixed_final(trial, input_size, output_size, 
                                                               final_hidden_layer_size, 
                                                               n_layers, p, out_fs,
                                                               max_neurons_layers=500)
                
            elif model_type == 'dynamic':
                out_fs = [trial.params[f'n_units_l{i}'] for i in range(n_layers)]
                model = architecture.dynamic_model(trial, input_size, output_size,
                                                   n_layers, p, out_fs,
                                                   max_neurons_layers=500)   

            else:
                print('not a valid model')

            # print('N BSQ: ', str(n_BSQ), 'N nwLH: ', str(n_nwLH), 'layer sizes: ', out_fs)
            # print('N BSQ: ', str(n_BSQ), 'N nwLH: ', str(n_nwLH), 'n layers: ', n_layers)
    
            
            model.to(device)  
            
            # load best-model, if it exists
            if os.path.exists(fmodel):  
                print('Loading model...')
                model.load_state_dict(torch.load(fmodel, map_location=torch.device(device)))
            else:
                raise Exception('model doesnt exists!!!')
            
            
            # define loss function
            criterion = nn.MSELoss() 
            
            # get the data
            test_loader = data.create_dataset(mode, seed, f_Pk, f_Pk_norm, f_params, 
                                              batch_size, shuffle=False, workers=1, 
                                              cosm_type =cosm_type, log=log, 
                                              shuffle_all = True)
            test_points = 0
            for x,y in test_loader:  test_points += x.shape[0]
            
            # define the arrays containing the true and predicted value of the parameters
            params  = output_size
            results = np.zeros((test_points, 2*params), dtype=np.float32)
            
            # test the model
            test_loss, points = 0.0, 0
            model.eval()
            with torch.no_grad():
                for x, y in test_loader:
                    bs   = x.shape[0]  #batch size
                    x, y = x.to(device), y.to(device)
                    y_NN = model(x)
                    test_loss += (criterion(y_NN, y).item())*bs
                    results[points:points+bs,0*params:1*params] = y.cpu().numpy()
                    results[points:points+bs,1*params:2*params] = y_NN.cpu().numpy()
                    points    += bs
            test_loss /= points
            print('Test loss:', test_loss)
    
            np.savetxt(fout, results)
```

```python
def plot_mse(
    Pk_type,
    cosm_type,
    n_sims_cosm,
    n_sims_BSQ,
    n_best_models,
    include_dashed_BSQ,
    mode = 'test'):
    
    if cosm_type == 'nwLH':
        minimum = np.array([0.1, 0.02, 0.50, 0.80, 0.60, 0.01])
        maximum = np.array([0.5, 0.08, 0.90, 1.20, 1.00, 1.0])
        param_labels = [r'$\Omega_m$', r'$\Omega_b$', r'$h$', r'$n_s$', r'$\sigma_8$', r'$M_{\nu}$']
        num_params = 6
        if Pk_type == 'MPk':
            title = 'Fine Tuning for Massive Neutrino (MP)'
        else:
            title = 'Fine Tuning for Massive Neutrino (P)'
    
    elif cosm_type == 'LC':
        minimum = np.array([0.1, 0.02, 0.50, 0.80, 0.60, -300])
        maximum = np.array([0.5, 0.08, 0.90, 1.20, 1.00, 300])
        param_labels = [r'$\Omega_m$', r'$\Omega_b$', r'$h$', r'$n_s$', r'$\sigma_8$', r'$f_{NL}^{local}$']
        num_params = 6
        title = 'Fine Tuning for Primordial non-Gaussianities (P)'
        
    elif cosm_type == 'EQ':
        minimum = np.array([0.1, 0.02, 0.50, 0.80, 0.60, -600])
        maximum = np.array([0.5, 0.08, 0.90, 1.20, 1.00, 600])
        param_labels = [r'$\Omega_m$', r'$\Omega_b$', r'$h$', r'$n_s$', r'$\sigma_8$', r'$f_{NL}^{equilateral}$']
        num_params = 6
        title = 'Fine Tuning for Primordial non-Gaussianities (P)'
    
    elif cosm_type == 'fR':
        minimum = np.array([0.1, 0.02, 0.50, 0.80, 0.60, 0.01, -3*10**(-4)])
        maximum = np.array([0.5, 0.08, 0.90, 1.20, 1.00, 1, 0])
        param_labels = [r'$\Omega_m$', r'$\Omega_b$', r'$h$', r'$n_s$', r'$\sigma_8(GR)$', r'$M_{\nu}$', r'$f_{RO}$']
        num_params = 7
        title = 'Fine Tuning for Modified Gravity (P)'
    
    else:
        print ('Cosm type?')
    
    colors = ['black', 'cornflowerblue', 'mediumturquoise', 'yellowgreen', 'sandybrown', 'red']
    
    fig, axes = plt.subplots(num_params+1, 1, figsize=(15, 65), sharex = True)
    
    for color_idx, n_BSQ in enumerate(n_sims_BSQ):
        all_mse_avg = []  # shape will be (len(n_sims_nwLH), num_params)
        all_mse_avg_param_sum_unnorm = []
        all_mse_std = []  # shape will be (len(n_sims_nwLH), num_params)
        all_mse_std_param_sum_unnorm = []
    
    
        for n_cosm in n_sims_cosm:
            mse_all_models = []
            mse_all_models_unnorm = []
    
            for n_model in range(n_best_models):
                # if n_BSQ == 0 and cosm_type == 'nwLH' and Pk_type == 'MPk':
                #     additional_extension = 'v2'
                    # params_ext = ''
                if cosm_type == 'fR':
                    params_ext = '_lcdm'
                    additional_extension = '_fixed'
                else:
                    additional_extension = '_fixed'
                    params_ext = ''
                    
                name = 'transfer10_network2_'+str(n_cosm)+'_'+cosm_type+'_'+str(n_BSQ)+'_BSQ'+additional_extension
                test = np.loadtxt('results/Results_'+mode+'_'+str(Pk_type)+'_'+str(cosm_type)+'_params_'+str(name)+params_ext+'_model'+str(n_model)+'.txt')
    
                test_true = test[:, :num_params]
                test_pred = test[:, num_params:]
                test_true_norm = test_true * (maximum - minimum) + minimum
                test_pred_norm = test_pred * (maximum - minimum) + minimum
    
                mse = [mean_squared_error(test_true_norm[:, i], test_pred_norm[:, i]) for i in range(num_params)]  #size 6
                mse_unnorm = [mean_squared_error(test_true[:, i], test_pred[:, i]) for i in range(num_params)]
                mse_all_models.append(mse)  #size 10 x 6
                mse_all_models_unnorm.append(mse_unnorm)
    
            mse_all_models = np.array(mse_all_models)  # shape: (10, 6)
            all_mse_avg.append(np.mean(mse_all_models, axis=0))  # (6,)
            all_mse_std.append(np.std(mse_all_models, axis=0))  # (6,)
            
            mse_all_models_unnorm = np.array(mse_all_models_unnorm)
            mse_param_sum_unnorm = np.sum(mse_all_models_unnorm, axis=1)  # shape: (10,)
            all_mse_avg_param_sum_unnorm.append(np.mean(mse_param_sum_unnorm))
            all_mse_std_param_sum_unnorm.append(np.std(mse_param_sum_unnorm))
    
        
        all_mse_avg = np.array(all_mse_avg)  # 5 n_sims_nwLH x 6 num_params
        all_mse_avg_param_sum_unnorm = np.array(all_mse_avg_param_sum_unnorm)
        all_mse_std = np.array(all_mse_std)  # 5 n_sims_nwLH x 6 num_params
        all_mse_std_param_sum_unnorm = np.array(all_mse_std_param_sum_unnorm)
    
        for param_idx in range(num_params):
            ax = axes[param_idx]
            if param_idx == 0:
                ax.plot(n_sims_cosm, all_mse_avg[:, param_idx], 
                        color=colors[color_idx], marker='o', label=r'$N_{\Lambda {\rm CDM}}$ = '+str(n_BSQ))
            else: 
                ax.plot(n_sims_cosm, all_mse_avg[:, param_idx], 
                        color=colors[color_idx], marker='o')
            
            ax.fill_between(n_sims_cosm,
                    all_mse_avg[:, param_idx] - all_mse_std[:, param_idx],
                    all_mse_avg[:, param_idx] + all_mse_std[:, param_idx],
                    color=colors[color_idx],
                    alpha=0.2)
            
            ax.set_ylabel('MSE ('+param_labels[param_idx]+')', fontsize = 50, labelpad=20)
            ax.set_xscale('log')
            ax.set_xticks([200, 1e3])
            ax.tick_params(axis='both', labelsize=30)
            
            
        
        axes[num_params].plot(n_sims_cosm, all_mse_avg_param_sum_unnorm, color = colors[color_idx],  
                     marker='o', label=r'$N_{\Lambda {\rm CDM}}$ = '+str(n_BSQ))
        axes[num_params].fill_between(n_sims_cosm,
                    all_mse_avg_param_sum_unnorm - all_mse_std_param_sum_unnorm,
                    all_mse_avg_param_sum_unnorm + all_mse_std_param_sum_unnorm,
                    color=colors[color_idx],
                    alpha=0.2)
        axes[num_params].set_ylabel('MSE (sum over parameters)',  fontsize = 40, labelpad=20)
        axes[num_params].set_xlabel(r'$N_{\rm fine-tune}}$', fontsize = 50, labelpad=15)
        axes[num_params].set_xscale('log')
        axes[num_params].set_xticks([200, 1e3])
        axes[num_params].set_xticklabels([r'$200$', r'$10^3$'])
        axes[num_params].tick_params(axis='both', labelsize=30)
        axes[num_params].tick_params(axis='x', labelsize=40)

        
    
    
    
    ## First network
    if include_dashed_BSQ == True:
        all_mse = []
        for n_BSQ in [2000, 4000, 8000, 16000, 22000]: 
            # name = 'transfer10_network1_0_nwLH_'+str(n_BSQ)+'_BSQ'
            name = 'transfer10_network1_'+str(n_BSQ)+'_BSQ'+additional_extension
            num_params = 5
            # additional_extension = ''
            Pk_type_BSQ = 'Pk'
            test = np.loadtxt('results/Results_test_'+str(Pk_type_BSQ)+'_BSQ_params_'+str(name)+'.txt')
            test_true = test[:, :num_params]
            test_pred = test[:, num_params:]
            minimum = np.array([0.1, 0.02, 0.50, 0.80, 0.60])
            maximum = np.array([0.5, 0.08, 0.90, 1.20, 1.00])
            test_true = test_true * (maximum - minimum) + minimum
            test_pred = test_pred * (maximum - minimum) + minimum
            mse = [mean_squared_error(test_true[:, i], test_pred[:, i]) for i in range(num_params)]  #size 5, one for each param
            all_mse.append(mse)  # 5 n sims x 5 cosmo params
        
        all_mse = np.array(all_mse)
        for param_indx in range (5):
            ax = axes[param_indx]
            for BSQ_indx in range (5):
                ax.hlines(all_mse[BSQ_indx, param_indx], xmin=n_sims_cosm[0], xmax=n_sims_cosm[-1], linestyle = '--', color = colors[BSQ_indx+1])
        
    
    
        from matplotlib.lines import Line2D
        legend_elements = [
            Line2D([0], [0], color='black', linestyle='-', label='Transfer learning'),
            Line2D([0], [0], color='black', linestyle='--', label='First network'),
        ]
        axes[1].legend(handles=legend_elements, fontsize= 35)
    
    # axes[-1].legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    # for ax in axes:
        # ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    axes[0].legend(loc='upper right', fontsize = 30)
    
    plt.suptitle(title, fontsize = 50, y=0.96, x=0.55)
    plt.tight_layout(rect=[0.03, 0, 0.97, 0.96])
    # Force fixed height (e.g. 65 inches) and automatic width
    fig.set_size_inches(fig.get_size_inches()[0], 65)  # keep width, set height
    # plt.savefig('mse_'+cosm_type+'_'+Pk_type+'.png', bbox_inches='tight')
    plt.show()





```
