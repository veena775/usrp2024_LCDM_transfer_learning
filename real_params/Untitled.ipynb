```python
import numpy as np


original_data = np.loadtxt('LCLH_params_OG.txt')  # shape: (1000,)

# Make sure it's a 2D column for stacking
original_data = original_data.reshape(-1, 1)  # shape: (1000, 1)


prepend_values = np.array([0.3175, 0.049, 0.6711, 0.9624, 0.834])  # replace with actual values

prepended_array = np.tile(prepend_values, (original_data.shape[0], 1))  # shape: (1000, 5)

combined = np.hstack((prepended_array, original_data))  # shape: (1000, 6)

np.savetxt('LC_params.txt', combined, fmt='%.6f')
```
